```python
import cv2
import numpy
import matplotlib.pyplot as plt
from google.colab import files
```

```python
arquivo = files.upload()
```

```python
img_bgr = cv2.imread("look_inside.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_rgb
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(img_rgb)
plt.title('img rgb')
plt.axis('off')

plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(10, 10));

ax[0].imshow(img_rgb)
ax[0].set_title('img rgb')

ax[1].imshow(img_bgr)
ax[1].set_title('img bgr')

ax[2].imshow(img_hsv)
ax[2].set_title('img hsv')

plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 10));

ax[0, 0].imshow(img_rgb)
ax[0, 0].set_title('img rgb')
ax[0, 0].axis('off')

ax[0, 1].imshow(img_bgr)
ax[0, 1].set_title('img bgr')
ax[0, 1].axis('off')

ax[1, 0].imshow(img_hsv)
ax[1, 0].set_title('img hsv')
ax[1, 0].axis('off')

ax[1, 1].imshow(img_gray, cmap="gray")
ax[1, 1].set_title('img gray')
ax[1, 1].axis('off')

plt.tight_layout()
plt.savefig('look-inside-cat-joao-lira.jpg')
plt.show()
```
