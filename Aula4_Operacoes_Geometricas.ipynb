```python
img_bgr = cv2.imread('barraskewda.png')
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
altura = img_gray.shape[0]
largura = img_gray.shape[1]
print(f"Altura: {altura} | Largura: {largura}")
```

## Operação Geométrica

```python
dx = 50
dy = 30

transladada = np.zeros((altura, largura), dtype=np.uint8)

transladada
```

## Operação Geométrica # 1: Translação de Exclusão

```python
for x in range (largura):
  for y in range (altura):
    novo_x = x + dx
    novo_y = y + dy

    if novo_x > 0 and novo_x < largura:
      if novo_y > 0 and novo_y < largura:
        transladada[novo_y, novo_x] = img_gray[y, x]

transladada
```

```python
plt.figure(figsize=(7,5))
plt.imshow(transladada, cmap='gray')
plt.title('Transladação de Exclusão')
plt.axis('off')
plt.show()
```

## Operação Geométrica # 2: Translação de Wrap-Around

```python
dx = 80
dy = 90

for x in range (largura):
  for y in range (altura):
    novo_x = (x + dx) % largura
    novo_y = (y + dy) % altura

    transladada[novo_y, novo_x] = img_gray[y, x]


plt.figure(figsize=(7,5))
plt.imshow(transladada, cmap='gray')
plt.title('Transladação de Wrap-Around')
plt.axis('off')
plt.show()
```

## Operação Geométrica # 3: Escalamento

```python
from matplotlib.image import imsave
sx = 10
sy = 10

nova_largura = int(largura * sx)
nova_altura = int(altura * sy)

escalada = np.zeros((nova_altura, nova_largura), dtype=np.uint8)

for x in range(nova_largura):
  for y in range (nova_altura):

    novo_x = int(x / sx)
    novo_y = int(y / sy)

    escalada[y, x] = img_gray[novo_y, novo_x]

plt.figure(figsize=(7,5))
#plt.imsave('barraskewda')
plt.imshow(escalada, cmap='gray')
plt.title('Escalamento')
plt.axis('off')
plt.show()
```

##Operação Geométrica #4: Espelhamento

```python
espelhada = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):

    novo_x = largura - 1 - x

    espelhada[y, x] = img_gray[y, novo_x]

espelhada

plt.figure(figsize=(7,5))
plt.imshow(espelhada, cmap='gray')
plt.title('Espelhamento')
plt.axis('off')
plt.show()
```
